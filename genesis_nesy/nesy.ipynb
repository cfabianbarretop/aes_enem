```python
import torch
import scallopy
```

```python
# Crear un contexto de Scallop
ctx = scallopy.ScallopContext(provenance="difftopkproofs")

# Definir relaciones
ctx.add_relation("digit_1", int)
ctx.add_relation("digit_2", int)

prob_digit_1 = [
    (0.05, (0,)),
    (0.10, (1,)),
    (0.15, (2,)),
    (0.10, (3,)),
    (0.20, (4,)),
    (0.10, (5,)),
    (0.05, (6,)),
    (0.10, (7,)),
    (0.05, (8,)),
    (0.10, (9,))
]

prob_digit_2 = [
    (0.10, (0,)),
    (0.05, (1,)),
    (0.10, (2,)),
    (0.15, (3,)),
    (0.10, (4,)),
    (0.15, (5,)),
    (0.05, (6,)),
    (0.10, (7,)),
    (0.10, (8,)),
    (0.10, (9,))
]

# Agregar hechos probabilísticos
ctx.add_facts("digit_1", prob_digit_1)  # hechos simples
ctx.add_facts("digit_2", prob_digit_2)

# Definir regla de suma
ctx.add_rule("sum(a + b) :- digit_1(a), digit_2(b)")

# Ejecutar el programa
ctx.run()
```

```python

# Probabilidades de digit_1
p1 = torch.tensor([
    0.05, 0.10, 0.15, 0.10, 0.20,
    0.10, 0.05, 0.10, 0.05, 0.10
])

# Probabilidades de digit_2
p2 = torch.tensor([
    0.10, 0.05, 0.10, 0.15, 0.10,
    0.15, 0.05, 0.10, 0.10, 0.10
])


# Guardar probabilidades de las sumas 0-18
sum_probs = torch.zeros(19)

for d1 in range(10):
    for d2 in range(10):

        suma = d1 + d2

        # probabilidad de esta prueba
        prob = p1[d1] * p2[d2]

        # acumular en la suma correspondiente
        sum_probs[suma] += prob
```

```python
# Consultar resultados
for prob, value in ctx.relation("sum"):
    print(f"{value[0]} -> {prob:.3f}")
```

```python
for i, prob in enumerate(sum_probs):
    print(f"{i} -> {prob:.4f}")
```
